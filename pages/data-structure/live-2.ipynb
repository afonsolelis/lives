## 1. A ideia de hierarquia

Arvores aparecem quando um elemento se conecta a outros em niveis:

- pastas e subpastas
- menus de navegacao
- chaves de campeonato
- regras de decisao
- indices de dados

A arvore binaria e a versao em que cada no tem no maximo **dois filhos**.

```python
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.esquerda = None
        self.direita = None

raiz = Node(50)
raiz.esquerda = Node(30)
raiz.direita = Node(70)
raiz.esquerda.esquerda = Node(20)
raiz.esquerda.direita = Node(40)
raiz.direita.esquerda = Node(60)
raiz.direita.direita = Node(80)

print('Arvore criada com raiz =', raiz.valor)
```

### O que faz dessa estrutura uma BST?

Uma BST (Arvore Binaria de Busca) segue a regra:

- tudo que esta na esquerda e menor que o no atual
- tudo que esta na direita e maior que o no atual

Essa regra permite decidir para qual lado andar durante a busca.

---

## 2. Inserindo valores

Vamos construir a BST como um sistema que recebe novas pontuacoes de jogadores.

```python
def inserir(no, valor):
    if no is None:
        return Node(valor)

    if valor < no.valor:
        no.esquerda = inserir(no.esquerda, valor)
    else:
        no.direita = inserir(no.direita, valor)

    return no

pontuacoes = [50, 30, 70, 20, 40, 60, 80, 65, 10, 35]
arvore = None
for p in pontuacoes:
    arvore = inserir(arvore, p)

print('BST criada com', len(pontuacoes), 'pontuacoes.')
```

### Situação real

Imagine um jogo que precisa responder rapidamente:

- existe alguem com score 65?
- qual o maior score do servidor?
- qual o score imediatamente acima de 40?

A BST nao resolve tudo sozinha, mas ajuda bastante na organizacao inicial.

---

## 3. Busca na BST

```python
def buscar(no, alvo):
    passos = []
    atual = no
    while atual is not None:
        passos.append(atual.valor)
        if alvo == atual.valor:
            return True, passos
        if alvo < atual.valor:
            atual = atual.esquerda
        else:
            atual = atual.direita
    return False, passos

encontrou, passos = buscar(arvore, 65)
print('Encontrou?', encontrou)
print('Caminho percorrido:', passos)
```

Essa e a grande ideia da BST: em vez de olhar todos os elementos, vamos descartando metade das possibilidades a cada comparacao, se a arvore estiver bem estruturada.

---

## 4. Travessias

Cada travessia responde a uma necessidade diferente.

```python
def inorder(no, saida=None):
    if saida is None:
        saida = []
    if no:
        inorder(no.esquerda, saida)
        saida.append(no.valor)
        inorder(no.direita, saida)
    return saida

def preorder(no, saida=None):
    if saida is None:
        saida = []
    if no:
        saida.append(no.valor)
        preorder(no.esquerda, saida)
        preorder(no.direita, saida)
    return saida

def postorder(no, saida=None):
    if saida is None:
        saida = []
    if no:
        postorder(no.esquerda, saida)
        postorder(no.direita, saida)
        saida.append(no.valor)
    return saida

print('In-order :', inorder(arvore))
print('Pre-order:', preorder(arvore))
print('Post-order:', postorder(arvore))
```

### Observacao importante

Em uma BST, a travessia **in-order** devolve os valores em ordem crescente. Isso e uma propriedade muito util.

## 5. Menor e maior valor

Em muitos sistemas reais, queremos extremos:

- produto mais barato
- maior pontuacao
- menor ID disponivel
- timestamp mais antigo

```python
def minimo(no):
    atual = no
    while atual.esquerda is not None:
        atual = atual.esquerda
    return atual.valor

def maximo(no):
    atual = no
    while atual.direita is not None:
        atual = atual.direita
    return atual.valor

print('Menor valor:', minimo(arvore))
print('Maior valor:', maximo(arvore))
```

## 6. O problema da arvore desbalanceada

Se inserirmos dados em ordem crescente, a BST pode virar quase uma lista ligada.

```python
arvore_torta = None
for valor in [10, 20, 30, 40, 50, 60, 70]:
    arvore_torta = inserir(arvore_torta, valor)

print('In-order da arvore torta:', inorder(arvore_torta))
print('A estrutura esta correta, mas a altura cresce demais.')
```

### Impacto em producao

Um indice que deveria responder rapidamente pode ficar lento se a estrutura perder o equilibrio. E por isso que arvores balanceadas como AVL e Red-Black existem.
