## 📖 Capítulo 1: O Problema do Estoque

### Contexto

Você é desenvolvedor em um e-commerce e precisa implementar um sistema para gerenciar o estoque de produtos. O primeiro requisito é simples: **armazenar uma lista de produtos** e acessar rapidamente pelo índice.

### 🤔 Qual estrutura usar?

````python
# Solução com ARRAY
# Arrays são perfeitos quando precisamos de acesso rápido por índice

estoque = [
    {"id": 1, "nome": "Notebook", "quantidade": 50},
    {"id": 2, "nome": "Mouse", "quantidade": 200},
    {"id": 3, "nome": "Teclado", "quantidade": 150},
    {"id": 4, "nome": "Monitor", "quantidade": 30},
    {"id": 5, "nome": "Headset", "quantidade": 80}
]

# Acesso O(1) - instantâneo!
print(f"Produto no índice 0: {estoque[0]['nome']}")
print(f"Quantidade do Monitor: {estoque[3]['quantidade']} unidades")
````

### ✅ Por que Array?

| Vantagem | Explicação |
|----------|------------|
| O(1) acesso | Acesso direto pelo índice |
| Simplicidade | Fácil de implementar |
| Memória contígua | Cache-friendly |

### ⚠️ Problema emergente...

O e-commerce cresceu! Agora precisamos **buscar produtos pelo ID** frequentemente. Com array, a busca é O(n)...

````python
# Busca linear em Array - O(n)
def buscar_produto_array(lista, produto_id):
    for produto in lista:
        if produto["id"] == produto_id:
            return produto
    return None

# Precisamos percorrer todo o array em pior caso!
produto = buscar_produto_array(estoque, 5)
print(f"Produto encontrado: {produto}")
````

## 📖 Capítulo 2: A Otimização com Hash Table

### 🎯 Novo requisito

O sistema está lento! Precisamos de buscas **mais rápidas** pelo ID do produto.

### 💡 Solução: Hash Table (Dicionário)

````python
# Refatorando com HASH TABLE
# Agora o acesso é O(1) usando a chave!

estoque_hash = {
    1: {"id": 1, "nome": "Notebook", "quantidade": 50},
    2: {"id": 2, "nome": "Mouse", "quantidade": 200},
    3: {"id": 3, "nome": "Teclado", "quantidade": 150},
    4: {"id": 4, "nome": "Monitor", "quantidade": 30},
    5: {"id": 5, "nome": "Headset", "quantidade": 80}
}

# Acesso O(1) - não importa o tamanho do estoque!
print(f"Busca por ID 5: {estoque_hash[5]['nome']}")
print(f"Busca por ID 2: {estoque_hash[2]['nome']}")

# Inserir novo produto também é O(1)!
estoque_hash[6] = {"id": 6, "nome": "Webcam", "quantidade": 45}
print(f"Novo produto: {estoque_hash[6]['nome']}")
````

### ✅ Comparação: Array vs Hash Table

| Operação | Array | Hash Table |
|----------|-------|------------|
| Acesso por índice | O(1) | N/A |
| Busca por ID | O(n) | **O(1)** |
| Inserção | O(n) | **O(1)** |
| Remoção | O(n) | **O(1)** |

### 🎓 Aprendizado

> **Hash Tables** são ideais quando precisamos de buscas frequentes por uma chave específica!

## 📖 Capítulo 3: O Sistema de Pedidos

### Contexto

Agora precisamos gerenciar a **fila de pedidos**. Os pedidos devem ser processados na ordem em que chegam (primeiro que entra, primeiro que sai).

### 🤔 Qual estrutura usar?

````python
from collections import deque

# FILA (Queue) - FIFO: First In, First Out
fila_pedidos = deque()

# Pedidos chegando...
fila_pedidos.append({"pedido_id": 101, "cliente": "Alice"})
fila_pedidos.append({"pedido_id": 102, "cliente": "Bob"})
fila_pedidos.append({"pedido_id": 103, "cliente": "Charlie"})

print(f"Fila atual: {list(fila_pedidos)}")

# Processando pedidos (sempre o mais antigo primeiro)
pedido_processado = fila_pedidos.popleft()
print(f"Processando: {pedido_processado}")
print(f"Fila restante: {list(fila_pedidos)}")
````

### ✅ Por que Fila (Queue)?

| Característica | Benefício |
|----------------|-----------|
| FIFO | Garante ordem justa de processamento |
| O(1) enqueue | Adicionar pedidos é rápido |
| O(1) dequeue | Processar pedidos é rápido |

### 🎯 Aplicações reais de Filas:
- Processamento de pedidos
- Filas de impressão
- Task queues (Celery, RabbitMQ)
- BFS (Busca em Largura)

## 📖 Capítulo 4: O Histórico de Navegação

### Contexto

Precisamos implementar o botão **"Voltar"** do site. O usuário deve voltar às páginas na ordem inversa de visita (última página visitada é a primeira a ser recuperada).

### 💡 Solução: Pilha (Stack)

````python
# PILHA (Stack) - LIFO: Last In, First Out
historico = []

# Usuário navegando...
historico.append("/home")           # Página inicial
historico.append("/produtos")       # Catálogo
historico.append("/produto/123")    # Detalhe do produto
historico.append("/carrinho")       # Carrinho

print(f"Histórico: {historico}")

# Usuário clica em "Voltar" 3 vezes
print(f"\nVoltando... {historico.pop()}")  # /carrinho
print(f"Voltando... {historico.pop()}")  # /produto/123
print(f"Voltando... {historico.pop()}")  # /produtos

print(f"\nPágina atual: {historico[-1]}")
````

### ✅ Por que Pilha (Stack)?

| Característica | Benefício |
|----------------|-----------|
| LIFO | Último a entrar é primeiro a sair |
| O(1) push | Adicionar páginas é rápido |
| O(1) pop | Voltar página é rápido |

### 🎯 Aplicações reais de Pilhas:
- Histórico de navegação
- Undo/Redo
- Call stack (funções recursivas)
- DFS (Busca em Profundidade)
- Validação de expressões

## 📖 Capítulo 5: A Árvore de Categorias

### Contexto

O e-commerce tem categorias e subcategorias de produtos organizadas hierarquicamente:

```
Eletrônicos
├── Informática
│   ├── Notebooks
│   └── Periféricos
└── Celulares
    ├── Smartphones
    └── Acessórios
```

### 🤔 Qual estrutura usar?

````python
# ÁRVORE (Tree) - Estrutura hierárquica

class CategoriaNode:
    def __init__(self, nome):
        self.nome = nome
        self.subcategorias = []

    def adicionar(self, subcategoria):
        self.subcategorias.append(subcategoria)

    def __repr__(self, level=0):
        indent = "  " * level
        result = f"{indent}📁 {self.nome}"
        for sub in self.subcategorias:
            result += "\n" + sub.__repr__(level + 1)
        return result

# Construindo a árvore
raiz = CategoriaNode("Eletrônicos")

informatica = CategoriaNode("Informática")
celulares = CategoriaNode("Celulares")

notebooks = CategoriaNode("Notebooks")
perifericos = CategoriaNode("Periféricos")
smartphones = CategoriaNode("Smartphones")
acessorios = CategoriaNode("Acessórios")

# Montando a hierarquia
informatica.adicionar(notebooks)
informatica.adicionar(perifericos)
celulares.adicionar(smartphones)
celulares.adicionar(acessorios)

raiz.adicionar(informatica)
raiz.adicionar(celulares)

print(raiz)
````

### Travessia em Árvore (DFS)

````python
# DFS - Depth First Search (Busca em Profundidade)
def buscar_categoria(node, nome_busca):
    """Busca uma categoria na árvore usando DFS"""
    if node.nome == nome_busca:
        return node

    for sub in node.subcategorias:
        resultado = buscar_categoria(sub, nome_busca)
        if resultado:
            return resultado

    return None

# Buscando "Smartphones"
resultado = buscar_categoria(raiz, "Smartphones")
print(f"Categoria encontrada: {resultado.nome if resultado else 'Não encontrada'}")

# Buscando "Livros" (não existe)
resultado = buscar_categoria(raiz, "Livros")
print(f"Categoria encontrada: {resultado.nome if resultado else 'Não encontrada'}")
````

### ✅ Por que Árvore?

| Característica | Benefício |
|----------------|-----------|
| Hierarquia | Representa relações pai-filho naturalmente |
| DFS/BFS | Algoritmos eficientes de travessia |
| Busca | O(log n) em árvores balanceadas |

### 🎯 Aplicações reais de Árvores:
- Estruturas hierárquicas (categorias, organogramas)
- Sistema de arquivos
- DOM do HTML
- Árvores de decisão
- Binary Search Trees (BST)

## 📖 Capítulo 6: O Sistema de Prioridades

### Contexto

Agora precisamos de um sistema de **suporte prioritário**. Clientes VIP devem ser atendidos primeiro, independentemente de quando chegaram.

### 💡 Solução: Heap (Priority Queue)

````python
import heapq

# HEAP - Min Heap (menor prioridade = atendido primeiro)
# Vamos usar prioridade negativa para criar um Max Heap

fila_prioridade = []

# Clientes chegando (prioridade, nome, problema)
# Prioridade: 1 = Baixa, 5 = Alta
clientes = [
    (3, "Cliente Normal A", "Dúvida sobre produto"),
    (5, "Cliente VIP X", "Problema crítico"),
    (1, "Cliente Normal B", "Sugestão"),
    (5, "Cliente VIP Y", "Problema de pagamento"),
    (2, "Cliente Normal C", "Reclamação"),
]

# Adicionando à heap (usamos prioridade negativa para max heap)
for prioridade, nome, problema in clientes:
    heapq.heappush(fila_prioridade, (-prioridade, nome, problema))

print("=== Fila de Prioridade ===")
print(f"Total de clientes: {len(fila_prioridade)}")

print("\n=== Atendendo por Prioridade ===")
while fila_prioridade:
    prioridade, nome, problema = heapq.heappop(fila_prioridade)
    print(f"[Prioridade {-prioridade}] {nome}: {problema}")
````

### ✅ Por que Heap?

| Característica | Benefício |
|----------------|-----------|
| O(1) acesso ao topo | Sempre acessa o elemento de maior prioridade |
| O(log n) inserção | Adicionar elementos é eficiente |
| O(log n) remoção | Remover o topo é eficiente |

### 🎯 Aplicações reais de Heaps:
- Priority queues
- Algoritmo de Dijkstra (menor caminho)
- Heap Sort
- Encontrar k-ésimo maior/menor elemento
- Agendamento de tarefas

## 📖 Capítulo 7: O Sistema de Recomendação

### Contexto

Precisamos implementar um sistema que recomenda produtos relacionados. Produtos são conectados por relações:

- Quem comprou X, também comprou Y
- Produtos similares
- Acessórios compatíveis

### 💡 Solução: Grafo (Graph)

````python
# GRAFO - Lista de Adjacência
# Cada produto é um nó, cada relação é uma aresta

from collections import defaultdict, deque

class GrafoProdutos:
    def __init__(self):
        self.grafo = defaultdict(list)

    def adicionar_relacao(self, produto1, produto2, tipo="similar"):
        """Adiciona uma relação entre produtos (grafo não-direcionado)"""
        self.grafo[produto1].append((produto2, tipo))
        self.grafo[produto2].append((produto1, tipo))

    def obter_relacionados(self, produto, max_relacionados=5):
        """Retorna produtos relacionados usando BFS"""
        if produto not in self.grafo:
            return []

        relacionados = []
        visitados = {produto}
        fila = deque([produto])

        while fila and len(relacionados) < max_relacionados:
            atual = fila.popleft()

            for vizinho, tipo in self.grafo[atual]:
                if vizinho not in visitados:
                    visitados.add(vizinho)
                    relacionados.append((vizinho, tipo))
                    fila.append(vizinho)

        return relacionados

    def mostrar_grafo(self):
        """Visualiza o grafo"""
        for produto, relacoes in self.grafo.items():
            vizinhos = ", ".join([f"{v}({t})" for v, t in relacoes])
            print(f"{produto} → {vizinhos}")
````

````python
# Criando o grafo de recomendações
grafo = GrafoProdutos()

# Adicionando relações
grafo.adicionar_relacao("Notebook Dell", "Mouse Sem Fio", "acessório")
grafo.adicionar_relacao("Notebook Dell", "Teclado Mecânico", "acessório")
grafo.adicionar_relacao("Notebook Dell", "Notebook HP", "similar")
grafo.adicionar_relacao("Notebook Dell", "Notebook Lenovo", "similar")
grafo.adicionar_relacao("Mouse Sem Fio", "Mousepad", "acessório")
grafo.adicionar_relacao("Teclado Mecânico", "Mousepad", "acessório")
grafo.adicionar_relacao("Notebook HP", "Notebook Lenovo", "similar")
grafo.adicionar_relacao("Notebook HP", "Mouse Sem Fio", "acessório")

print("=== Grafo de Relacionamentos ===")
grafo.mostrar_grafo()

print("\n=== Recomendações para 'Notebook Dell' ===")
relacionados = grafo.obter_relacionados("Notebook Dell")
for produto, tipo in relacionados:
    print(f"  📌 {produto} [{tipo}]")
````

### Busca em Profundidade (DFS) no Grafo

````python
def dfs_produtos(grafo, inicio, visitados=None):
    """DFS para explorar todos os produtos conectados"""
    if visitados is None:
        visitados = set()

    visitados.add(inicio)
    print(f"Visitando: {inicio}")

    for vizinho, tipo in grafo.grafo[inicio]:
        if vizinho not in visitados:
            dfs_produtos(grafo, vizinho, visitados)

    return visitados

print("=== DFS a partir de 'Notebook Dell' ===")
dfs_produtos(grafo, "Notebook Dell")
````

### ✅ Por que Grafo?

| Característica | Benefício |
|----------------|-----------|
| Relações complexas | Modela conexões múltiplas naturalmente |
| BFS/DFS | Algoritmos eficientes de travessia |
| Caminhos | Encontra conexões entre elementos |

### 🎯 Aplicações reais de Grafos:
- Redes sociais (amigos, seguidores)
- Sistemas de recomendação
- Mapas e rotas (GPS)
- Dependências entre pacotes
- PageRank do Google

## 📖 Capítulo 8: Desafio Final - Integrando Tudo

### 🎯 Problema Completo

Agora vamos integrar todas as estruturas em um **sistema completo de e-commerce**:

1. **Hash Table**: Catálogo de produtos (busca rápida por ID)
2. **Árvore**: Categorias hierárquicas
3. **Fila**: Processamento de pedidos
4. **Pilha**: Histórico de navegação
5. **Heap**: Suporte prioritário
6. **Grafo**: Sistema de recomendação

````python
# SISTEMA COMPLETO DE E-COMMERCE

class ECommerceSystem:
    def __init__(self):
        # Hash Table para produtos
        self.produtos = {}

        # Árvore para categorias
        self.categorias = {}

        # Fila para pedidos
        self.fila_pedidos = deque()

        # Pilha para histórico
        self.historico = []

        # Heap para suporte
        self.suporte = []

        # Grafo para recomendações
        self.recomendacoes = GrafoProdutos()

    def adicionar_produto(self, id_prod, nome, categoria, preco):
        self.produtos[id_prod] = {
            "id": id_prod,
            "nome": nome,
            "categoria": categoria,
            "preco": preco
        }
        print(f"✅ Produto adicionado: {nome}")

    def buscar_produto(self, id_prod):
        return self.produtos.get(id_prod)

    def fazer_pedido(self, cliente, itens):
        pedido = {"cliente": cliente, "itens": itens}
        self.fila_pedidos.append(pedido)
        print(f"🛒 Pedido de {cliente} adicionado à fila")

    def processar_pedido(self):
        if self.fila_pedidos:
            pedido = self.fila_pedidos.popleft()
            print(f"📦 Processando pedido de {pedido['cliente']}")
            return pedido
        print("Nenhum pedido na fila")
        return None

    def navegar(self, pagina):
        self.historico.append(pagina)
        print(f"🌐 Navegando para: {pagina}")

    def voltar(self):
        if self.historico:
            pagina = self.historico.pop()
            print(f"⬅️ Voltando de: {pagina}")
            return pagina
        return None

    def adicionar_suporte(self, prioridade, cliente, problema):
        heapq.heappush(self.suporte, (-prioridade, cliente, problema))
        print(f"🎫 Suporte adicionado: {cliente} (prioridade {prioridade})")

    def atender_suporte(self):
        if self.suporte:
            prioridade, cliente, problema = heapq.heappop(self.suporte)
            print(f"👤 Atendendo {cliente}: {problema}")
            return (cliente, problema)
        print("Nenhum cliente na fila de suporte")
        return None

    def relacionar_produtos(self, id1, id2, tipo="similar"):
        prod1 = self.produtos.get(id1, {}).get("nome", f"Produto {id1}")
        prod2 = self.produtos.get(id2, {}).get("nome", f"Produto {id2}")
        self.recomendacoes.adicionar_relacao(prod1, prod2, tipo)
        print(f"🔗 Relacionando {prod1} com {prod2}")

    def mostrar_recomendacoes(self, id_prod):
        nome = self.produtos.get(id_prod, {}).get("nome")
        if nome:
            relacionados = self.recomendacoes.obter_relacionados(nome)
            print(f"\n💡 Recomendações para {nome}:")
            for prod, tipo in relacionados:
                print(f"   📌 {prod} [{tipo}]")
            return relacionados
        return []
````

````python
# === DEMONSTRAÇÃO DO SISTEMA ===

ecommerce = ECommerceSystem()

# 1. Adicionando produtos (Hash Table)
print("=== 1. CADASTRO DE PRODUTOS ===")
ecommerce.adicionar_produto(1, "Notebook Dell", "Informática", 4500)
ecommerce.adicionar_produto(2, "Mouse Sem Fio", "Periféricos", 150)
ecommerce.adicionar_produto(3, "Teclado Mecânico", "Periféricos", 400)
ecommerce.adicionar_produto(4, "Notebook HP", "Informática", 4200)
ecommerce.adicionar_produto(5, "Monitor 24\"", "Periféricos", 900)

# Busca rápida (O(1))
print(f"\nBusca por ID 3: {ecommerce.buscar_produto(3)}")
````

````python
# 2. Relacionando produtos (Grafo)
print("\n=== 2. RELACIONAMENTOS ===")
ecommerce.relacionar_produtos(1, 2, "acessório")
ecommerce.relacionar_produtos(1, 3, "acessório")
ecommerce.relacionar_produtos(1, 4, "similar")
ecommerce.relacionar_produtos(2, 3, "complementar")
ecommerce.relacionar_produtos(4, 5, "acessório")

ecommerce.mostrar_recomendacoes(1)
````

````python
# 3. Navegação e Histórico (Pilha)
print("\n=== 3. NAVEGAÇÃO ===")
ecommerce.navegar("/home")
ecommerce.navegar("/informatica")
ecommerce.navegar("/produto/1")
ecommerce.navegar("/carrinho")

print("\nVoltando páginas:")
ecommerce.voltar()
ecommerce.voltar()
ecommerce.voltar()
````

````python
# 4. Pedidos (Fila)
print("\n=== 4. PEDIDOS ===")
ecommerce.fazer_pedido("Alice", [(1, 1), (2, 2)])
ecommerce.fazer_pedido("Bob", [(4, 1)])
ecommerce.fazer_pedido("Charlie", [(3, 1), (5, 1)])

print("\nProcessando pedidos:")
ecommerce.processar_pedido()
ecommerce.processar_pedido()
ecommerce.processar_pedido()
````

````python
# 5. Suporte Prioritário (Heap)
print("\n=== 5. SUPORTE ===")
ecommerce.adicionar_suporte(2, "Carlos", "Produto com defeito")
ecommerce.adicionar_suporte(5, "Diana VIP", "Urgente - Reembolso")
ecommerce.adicionar_suporte(1, "Eduardo", "Dúvida sobre frete")
ecommerce.adicionar_suporte(5, "Fernando VIP", "Problema crítico")
ecommerce.adicionar_suporte(3, "Gabriela", "Troca de produto")

print("\nAtendendo por prioridade:")
for _ in range(5):
    ecommerce.atender_suporte()
````

### Exercício 1: Validador de Parênteses
Use uma **pilha** para validar se uma expressão matemática tem parênteses balanceados.

````python
def validar_parenteses(expressao):
    """
    Valida se os parênteses estão balanceados usando pilha.
    Ex: "(a + b) * (c - d)" → True
    Ex: "(a + b * (c - d)" → False
    """
    pilha = []

    for char in expressao:
        if char == '(':
            pilha.append(char)
        elif char == ')':
            if not pilha:
                return False
            pilha.pop()

    return len(pilha) == 0

# Testes
print(validar_parenteses("(a + b) * (c - d)"))  # True
print(validar_parenteses("(a + b * (c - d)"))   # False
print(validar_parenteses("((()))"))             # True
print(validar_parenteses("(()))"))              # False
````

### Exercício 2: Sistema de Cache LRU
Implemente um cache **LRU (Least Recently Used)** usando Hash Table + Lista Duplamente Ligada.

````python
class Node:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None

class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.cache = {}  # Hash Table
        self.head = Node(0, 0)  # Dummy head
        self.tail = Node(0, 0)  # Dummy tail
        self.head.next = self.tail
        self.tail.prev = self.head

    def _add_node(self, node):
        """Adiciona nodo após o head (mais recente)"""
        node.prev = self.head
        node.next = self.head.next
        self.head.next.prev = node
        self.head.next = node

    def _remove_node(self, node):
        """Remove um nodo da lista"""
        node.prev.next = node.next
        node.next.prev = node.prev

    def _move_to_front(self, node):
        """Move nodo para o início (mais recente)"""
        self._remove_node(node)
        self._add_node(node)

    def get(self, key):
        if key in self.cache:
            node = self.cache[key]
            self._move_to_front(node)
            return node.value
        return -1

    def put(self, key, value):
        if key in self.cache:
            node = self.cache[key]
            node.value = value
            self._move_to_front(node)
        else:
            node = Node(key, value)
            self.cache[key] = node
            self._add_node(node)

            if len(self.cache) > self.capacity:
                # Remove o menos recente (antes do tail)
                lru = self.tail.prev
                self._remove_node(lru)
                del self.cache[lru.key]

# Teste
cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
print(cache.get(1))      # 1
cache.put(3, 3)          # Evicta chave 2
print(cache.get(2))      # -1
print(cache.get(3))      # 3
````

### Exercício 3: Árvore Binária de Busca
Implemente uma BST com inserção, busca e travessia.

````python
class BSTNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        if not self.root:
            self.root = BSTNode(value)
        else:
            self._insert_recursive(self.root, value)

    def _insert_recursive(self, node, value):
        if value < node.value:
            if node.left:
                self._insert_recursive(node.left, value)
            else:
                node.left = BSTNode(value)
        else:
            if node.right:
                self._insert_recursive(node.right, value)
            else:
                node.right = BSTNode(value)

    def search(self, value):
        return self._search_recursive(self.root, value)

    def _search_recursive(self, node, value):
        if not node:
            return False
        if node.value == value:
            return True
        if value < node.value:
            return self._search_recursive(node.left, value)
        return self._search_recursive(node.right, value)

    def inorder(self):
        """Travessia in-order (retorna elementos ordenados)"""
        result = []
        self._inorder_recursive(self.root, result)
        return result

    def _inorder_recursive(self, node, result):
        if node:
            self._inorder_recursive(node.left, result)
            result.append(node.value)
            self._inorder_recursive(node.right, result)

# Teste
bst = BinarySearchTree()
for val in [50, 30, 70, 20, 40, 60, 80]:
    bst.insert(val)

print(f"In-order: {bst.inorder()}")  # [20, 30, 40, 50, 60, 70, 80]
print(f"Buscar 40: {bst.search(40)}")  # True
print(f"Buscar 25: {bst.search(25)}")  # False
````
